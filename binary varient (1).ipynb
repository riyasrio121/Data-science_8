```python
penguins['bill_length_mm']=penguins['bill_length_mm'].fillna(penguins['bill_length_mm'].mean())
```

```python
penguins.isnull().sum()
```

```python
penguins['bill_depth_mm']=penguins['bill_depth_mm'].fillna(penguins['bill_depth_mm'].mean())
```

```python
penguins.isnull().sum()
```

```python
penguins['flipper_length_mm']=penguins['flipper_length_mm'].fillna(penguins['flipper_length_mm'].mean())
```

```python
penguins.isnull().sum()
```

```python
penguins['body_mass_g']=penguins['body_mass_g'].fillna(penguins['body_mass_g'].mean())
```

```python
penguins.isnull().sum()
```

```python
penguins['sex']=penguins['sex'].fillna(penguins['sex'].mode()[0])
```

```python
penguins.isnull().sum()
```

```python
penguins.duplicated()
```

```python
penguins.duplicated().sum()
```

```python
penguins=penguins.drop_duplicates()
```

```python
#SCATTER PLOT
```

```python
sns.scatterplot(data=penguins,x='flipper_length_mm',y='body_mass_g',hue='species')
plt.show()
```

```python
#VIOLIN PLOT
```

```python
sns.violinplot(data=penguins, x='flipper_length_mm', y='sex', hue='species')
plt.show()
```

```python
#BARPLOT
```

```python
sns.barplot(data=penguins,x='flipper_length_mm',y='sex')
plt.title("Barplot")
plt.show()
```

```python
#BOXPLOT
```

```python
sns.boxplot(data=penguins,y='bill_length_mm')
plt.title("Boxplot")
plt.show()
```
